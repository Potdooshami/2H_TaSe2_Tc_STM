```python
import numpy as np
to_idts = [0,5,3,1]
lbls = ['078K','110K','115K','118K']
clr_Ts = ["r","g","b","y"]
def set_xylim(ax,ind):
    xlim = np.array([0,512])+1000
    ylim = np.array([512,0])+1000
    if ind == 0:
        ax.set_xlim(xlim)
        ax.set_ylim(ylim)
```

### topo

```python
fm.id_block = 'topo'
fm.timestamp()
cmaps = ['gray','afmhot']
for ind,lbl,to_idt in zip(range(4),lbls,to_idts):
    for ic,cmap in zip(range(2),cmaps):
        id_file = '_'.join([lbl,cmap])
        print(to_idt,cmap)
        fig = plt.figure()
        ax = imgen.topo(to_idt,cmap=cmap)
        set_xylim(ax,ind)
        fig = ax.figure
        fig.savefig(fm.get_filepath(id_file),format='svg')
```
