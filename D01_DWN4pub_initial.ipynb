```python
idt = 0
fig = plt.figure(figsize=(5,5))

xlimylim_DVR = ((560,690),(1310,1440))
xlimylim_DVL = ((1860,1990),(260,390))
dw_r = (1142,1496)
dw_g= (1708,202)
dw_b = (1352,173)
```

```python
cw_vr = CropWindow()
cw_vr.set_xlimylim(*xlimylim_DVR)
cw_vl = CropWindow()
cw_vl.set_xlimylim(*xlimylim_DVL)
print(cw_vl,cw_vr)
cw_v = [cw_vr,cw_vl]

cw_r = CropWindow()
cw_r.set_by_anchor(dw_r,100)
cw_g = CropWindow()
cw_g.set_by_anchor(dw_g,100)
cw_b = CropWindow()
cw_b.set_by_anchor(dw_b,100)

cw_all = [cw_vr,cw_vl,cw_r,cw_g,cw_b]
```

```python
imgen.dw(idt)
from crypy_examples.dwn_toon import CLR_NODE

cw_vr.ax_cropbox(ls=':',ec=CLR_NODE[0,:])
cw_vl.ax_cropbox(ls=':',ec=CLR_NODE[1,:])
cw_r.ax_cropbox(ls='--',ec='r')
cw_g.ax_cropbox(ls='--',ec='g')
cw_b.ax_cropbox(ls='--',ec='b')
import matplotlib.pyplot as plt
plt.savefig('D01_DWN_Whole_v2.svg', format='svg', bbox_inches='tight')
```

```python
nms_obj = ['vor1','vor2','dw_r','dw_g','dw_b']
fncs = [lambda: imgen.topo(idt,cmap='gray'),
 lambda: imgen.dw(idt),
 lambda: imgen.vertex(idt)]
nms_fcn = ['topo','dw','vertex']
msb = 3
nsb = 5
fig,axs = plt.subplots(msb,nsb,figsize=(20,15))
for cw_v_fcs,nm_obj,jsb in zip(cw_all,nms_obj,range(nsb)):
    for fcn,nm_fcn,isb in zip(fncs,nms_fcn,range(msb)):        # fig = plt.figure(figsize=(5,5))
        plt.sca(axs[isb,jsb])
        fcn()
        # ax = plt.gca()
        # fullax()
        cw_v_fcs.ax_xylims()    
plt.savefig('D01_crop.svg', format='svg', bbox_inches='tight')
```
